```python
import pandas as pd
```

```python
df=pd.read_csv("sample_data.csv")
df
```

# use of groupby 

```python
avg_df=df.groupby('department')['salary'].mean()
```

# find duplicates

```python
df_duplicate=df[df.duplicated('city')]
df_duplicate
```

# drop duplicate

```python
df_duplicate.drop_duplicates(subset='city',inplace=True)
```

# filtering data

```python
df[(df['city']=='Delhi') & (df['age']>28)]
```

# handling with missing data

```python
df['city'].fillna(df['city'].mode().iloc[0])
```

# apply function

```python
df['New Salary']=df['salary'].apply(lambda x: x*1.10)
```
