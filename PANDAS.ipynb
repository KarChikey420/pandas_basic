```python
import pandas as pd
```

```python
df=pd.read_csv("sample_data.csv")
df
```

# use of groupby 

```python
avg_df=df.groupby('department')['salary'].mean()
```

# find duplicates

```python
df_duplicate=df[df.duplicated('city')]
df_duplicate
```

# drop duplicate

```python
df_duplicate.drop_duplicates(subset='city',inplace=True)
```

# filtering data

```python
df[(df['city']=='Delhi') & (df['age']>28)]
```

# handling with missing data

```python
df['city'].fillna(df['city'].mode().iloc[0])
```

# apply function

```python
df['New Salary']=df['salary'].apply(lambda x: x*1.10)
```

```python
df.head()
```

# sorting data

```python
df.sort_values(by='name',ascending=True,inplace=True)
df.head()
```

# Top N Records per Group

```python
df['rank'] = df.groupby('department')['salary'].rank(method='dense', ascending=False)
df[df['rank'] <= 2]
```

# Merge / Join DataFrames

```python
df1=pd.DataFrame({
    'id':[1,2,3],
    'name':['Alice','Bob','Charlie'],
    'department':['Hr','IT','Finance'],
    'salary':[50000,60000,55000]
})

df2=pd.DataFrame({
    'id':[4,5,6],
    'name':['David','Eve','Frank'],
    'department':['IT','Finance','Hr'],
    'salary':[65000,58000,52000]
})

merged_df=pd.concat([df1,df2],ignore_index=True)
```
