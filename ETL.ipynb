```python
import pandas as pd
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
def Extract():
    engine=create_engine(os.getenv('DATABASE_URL_1'))
    query=text("select * from employees")
    df=pd.read_sql(query,engine)
    return df.head()

print(Extract())
```
