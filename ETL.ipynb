```python
import pandas as pd
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
def Extract():
    engine=create_engine(os.getenv('DATABASE_URL_1'))
    query=text("select * from employees")
    df=pd.read_sql(query,engine)
    return df
```

```python
def Transform(df):
    avg_salary=df['salary'].mean()
    df['avg_salary']= df['salary'].apply(lambda x: avg_salary)
    df1=df.sort_values(by='salary',ascending=False)
    return df1
```

```python
def Load(df):
    engine=create_engine(os.getenv('DATABASE_URL_2'))
    df.to_sql('transformed_employees', engine, if_exists='replace', index=False)
```

```python
def main():
    df=Extract()
    print("Extracted Data...............:\n",df.head())
    
    trans_df=Transform(df)
    print("Transform data...............:\n",trans_df.head())
    
    load_df=Load(trans_df)
    print("Load data successfully....................:\n")
```

```python
if __name__ == "__main__":
    main()
```
